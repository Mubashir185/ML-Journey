```python
correct_pin = "4521"
while True:
    user_input = input("Enter your PIN (or type 'cancel' to exit): ")
    if user_input.lower() == "cancel":
        print("Transaction Cancelled.")
        break
    if user_input == correct_pin:
        print("Access Granted!")
        break
    else:
        print("Incorrect PIN, try again.")
```

```python
def convert_temperature(value, to_unit="Fahrenheit"):
    if to_unit == "Fahrenheit":
        return (value * 9/5) + 32
    elif to_unit == "Kelvin":
        return value + 273.15
    else:
        return "Invalid unit"
result_default = convert_temperature(25)
print(f"25°C converted using the default unit is: {result_default}°F")
result_positional = convert_temperature(25, "Kelvin")
print(f"25°C converted using a positional argument is: {result_positional}K")
result_keyword = convert_temperature(0, to_unit="Kelvin")
print(f"0°C converted using a keyword argument is: {result_keyword}K")
```

```python
CORRECT_ANSWERS = ["A", "C", "B", "D", "A"]
def grade_quiz(user_answers, correct_answers):
    attempted = len(user_answers)
    score = 0
    for i in range(attempted):
        if user_answers[i].strip().upper() == correct_answers[i]:
            score += 1
            
    percentage = (score / len(correct_answers)) * 100
    
    return {
        "score": score,
        "attempted": attempted,
        "percentage": round(percentage, 2)
    }
user_answers = []
current_index = 0
while current_index < len(CORRECT_ANSWERS):
    user_input = input(f"Question {current_index + 1} answer (or 'quit' to stop): ").strip()
    
    if user_input.lower() == "quit":
        print("\nTransaction Cancelled / Quiz stopped early.")
        break    
    user_answers.append(user_input)
    current_index += 1
quiz_results = grade_quiz(user_answers, CORRECT_ANSWERS)
print("\n--- Summary Results ---")
print(f"Correct:     {quiz_results['score']} / {len(CORRECT_ANSWERS)}")
print(f"Attempted:   {quiz_results['attempted']}")
print(f"Percentage:  {quiz_results['percentage']}%")
```
