```python
_num = int(input("Enter the Number: "))
check_num = _num
reversed_number = 0

while _num != 0:
    digit = _num % 10
    reversed_number = reversed_number * 10 + digit
    _num //= 10
print(reversed_number)
if(check_num == reversed_number):
    print("The Number is Palindrome")
else:
    print("The Number is not a Palindrome")
```

```python
import math

prime_num = int(input("Enter the Number: "))

# Step 1: Check if the number is less than 2
if prime_num < 2:
    print("Not a prime Number less than 2")

# Step 2: If it is 2 or bigger, check for divisors
else:
    limit = int(math.sqrt(prime_num))
    is_prime = True  # We assume it is prime until proven otherwise
    
    for i in range(2, limit + 1):
        if prime_num % i == 0:
            print("Not a prime number")
            is_prime = False
            break  # Stop checking since we found a divisor
            
    if is_prime:
        print("It is a prime number!")
```

Factorial Using Loops 

```python
n = int(input("Enter the Number: "))
num = n
fact = 1
if(n < 1):
    print(n)
else:
    while(n>1):
        fact *= n
        n -= 1
        
print("The Factorial of {} is {}".format(num,fact))
```

```python
def Factorial_recursion(fa):
    if(fa<1):
        return 1
    else:
        return fa * (Factorial_recursion(fa - 1) )

num = int(input("Enter the Number: "))
print("The Factorial of {} is ".format(num),Factorial_recursion(num))
```

```python
string_1 = input("Enter the Sentence: ")
vowels = "aeiouAEIOU"
vowel_count = 0
consonant_count = 0
digit_count = 0
space_count = 0
index = 0
while index < len(string_1):
    char = string_1[index]  
    
    if char.isdigit():      
        digit_count += 1
    elif char.isalpha():
        if char in vowels:
            vowel_count += 1
        else:
            consonant_count += 1
    elif char.isspace():
        space_count += 1
        
    index += 1 
print(string_1)
print(f"Vowels: {vowel_count} Consonants: {consonant_count} Digits: {digit_count} Spaces: {space_count}")
```

```python
def sum(n):
    if(n <= 0):
        return 0
    return n + sum(n -1)
num = int(input("Enter the Number: "))
print("Sum of Numbers till {} is {}".format(num , sum(num)))
```

```python
def power(base, exponent):
    if exponent == 0:
        return 1
    return base * power(base,exponent - 1)
ba = int(input("Enter the number: "))
expo = int(input("Enter the power: "))
print("The exponential value of base: {} with exponent: {} is {} ".format(ba,expo,power(ba,expo)))
```

```python
def fabonacci(n):
    if(n <= 1):
        return n
    return fabonacci(n - 1) + fabonacci(n - 2)
fab = int(input("Enter the Number: "))
print("The Fabonacii series till {} is: ".format(fab))
for i in range(fab):
    print(fabonacci(i), end=" ")
```
