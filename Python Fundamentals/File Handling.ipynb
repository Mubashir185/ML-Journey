```python
with open("practice.txt",'w') as f:
    f.write("Hi Everyone \n we are Learning file I/O \n using Java. \n I like Programming in Java.")
```

```python
present = False
with open("practice.txt","r+") as f:
    data = f.read()
    data_len = len(data) -4 
    learn = 'Learning' in data
    while( data_len >= 0):
        if data[data_len: data_len + 4] == "Java":
            data = data[:data_len] + "Python" + data[data_len + 4:]
        data_len -= 1
        f.seek(0)
        f.write(data)
        f.truncate()
    print(learn)
```

```python
with open("practice.txt","r") as f:
    data = f.read()
    lst = data.split(",")
    for num in lst:
        if int(num) % 2 == 0:
            print(num , "Even Number")
        else:
            print(num , "Odd Number")
        
```
