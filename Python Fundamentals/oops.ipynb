```python
class Student:
    # Default Constructor:
    def __init__(self):
            pass
    # Parameterized Constructor:
    def __init__(self,name,age,reg_id):
        self.full_name = name
        self.age = age
        self.roll_no = reg_id
name = input("Enter you Name: ")
age = int(input("Enter your Age: "))
roll_number = int(input("Enter your Registration Number: "))
s1 = Student(name,age,roll_number)
print("The Student Name: {} , age is {} , and Registration ID is {}".format(s1.full_name,s1.age,s1.roll_no))
```

```python
class student:
    def __init__(self , name,math_marks,physics_marks,chem_marks):
        self.std_name = name
        self.marks_1 = math_marks
        self.marks_2 = physics_marks
        self.marks_3 = chem_marks
    def average_marks(self):
        average = (self.marks_1 + self.marks_2 + self.marks_3) / 3
        print(average)
na = input("Enter the Name of Student: ")
sub_1 = int(input("Enter the Marks of Math: "))
sub_2 = int(input("Enter the Marks of Physics: "))
sub_3 = int(input("Enter the Marks of Chemistry: "))
s1 = student(na , sub_1 , sub_2,sub_3)
s1.average_marks()
```

```python
class Account:
    def __init__(self , bal , acc_no):
        self.balance = bal
        self.Account_num = acc_no
    def debit(self,amount):
        self.balance -= amount
    def credit(self,bo_amount):
        self.balance += bo_amount
    def dis_amount(self):
        print("---------Your Account Details---------")
        print("Your current Balance:",self.balance)
        print("Your Account Number: ",self.Account_num)

a1 = Account(1000,158742)
a1.debit(250)
a1.credit(1500)
a1.dis_amount()
```

```python
class Circle:
    def __init__(self , r):
        self.r = r
        self.pi = 3.14
    def area(self):
        area_cir = self.pi *(self.r * self.r)
        return area_cir
    def perimeter(self):
        perimeter_circle = 2 * (self.pi * self.r)
        return perimeter_circle
    def show_area(self):
        print("The Area of Circle is {}".format(self.area()))
        print("The Perimeter of Circle is {}".format(self.perimeter()))

c1 = Circle(4)
c1.show_area()
```

```python
class Empolyee:
    def __init__(self,role,salary,dept):
        self.role = role
        self.salary = salary
        self.dept = dept
    def showDetails(self):
        print("The Role of Employee is {}".format(self.role))
        print("The Salary of Employee is {}".format(self.salary))
        print("The Department of Employee is {}".format(self.dept))
class Engineer(Empolyee):
    def __init__(self,name,age,role,salary,dept):
        super().__init__(role,salary,dept)
        self.name = name
        self.age = age
    def showDetails(self):
        print("The Name of Employee is {}".format(self.name))
        print("The Age of Employee is {}".format(self.age))
        super().showDetails()
name = input("Enter the Name of Employee: ")
age = input("Enter the Age of Employee: ")
role = input("Enter the Role od Employee: ")
sal = input("Enter the Salary of Employee: ")
dept = input("Enter the Department of Employee: ")
eng_1 = Engineer(name,age,role,sal,dept)
eng_1.showDetails()
    
```

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance  

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if self.check_balance() >= amount:
            self.balance -= amount
            return self.balance
        else:
            return "Sorry for the inconvenience!\nInsufficient Balance for this Transaction!"

    def check_balance(self):
        return self.balance
initial_deposit = float(input("Enter the Initial Deposit Amount: "))
acc_1 = BankAccount(initial_deposit)
print("The Amount: {} is deposited into your account successfully!".format(initial_deposit))
print("The Balance available in your account is: {}\n".format(acc_1.check_balance()))
withdraw_amount = float(input("Enter the Amount to Withdraw: "))
withdrawal_result = acc_1.withdraw(withdraw_amount)
if isinstance(withdrawal_result, str):
    print(withdrawal_result)
else:
    print("The Amount {} was withdrawn successfully!".format(withdraw_amount))
    print("The remaining Balance available in your account is: {}".format(acc_1.check_balance()))
```

```python
class Student:

    def __init__(self,student_name,reg_id):
        self.name = student_name
        self.reg_id = reg_id
    def add_Student_marks(self,marks_list):
        self.marks = marks_list
        return "Marks added successfully!"
    def result(self):
        std_marks = sum(self.marks)
        total_marks = len(self.marks) * 100
        percentage = (std_marks / total_marks)*100
        if sum(self.marks) >= 40 and percentage >= 33:
            print("The Student {} with Registration ID {} has Passed the Exam with {}!".format(self.name,self.reg_id,percentage))

        else:
            print("The Student {} with Registration ID {} has Failed the Exam!".format(self.name,self.reg_id))

std_name = input("Enter the name of Student: ")
reg_id = input("Enter the Registration ID: ")           
std1 = Student(std_name,reg_id)
std1.add_Student_marks([80, 75, 90])
std1.result()
```
