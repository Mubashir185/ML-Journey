```python
def summarize(temps):
    dict_temp = {}
    if len(temps) == 0:
        return 
    else:
        min_temp = min(temps)
        max_temp = max(temps)
        avg_temp = sum(temps) / len(temps)
        dict_temp["minimum"] = min_temp
        dict_temp["maximum"] = max_temp     
        dict_temp["average"] = avg_temp
        return dict_temp
temperature = (input("Enter the temperature in Celsius: "))
temp_list = []
while temperature.lower() != "done" :
    if(float(temperature) < -273.15):
            continue
    else:
        temp_list.append(float(temperature))
        temperature = (input("Enter the temperature in Celsius: "))
print("List of valid temperatures:", temp_list)
summary = summarize(temp_list)
print("Summary of temperatures:", summary)
```

```python
shopping_list = [
    {"item": "Notebook", "price": 150, "purchased": False},
    {"item": "Pen", "price": 30, "purchased": True},
    {"item": "USB Drive", "price": 900, "purchased": False},
]
still_need = [item["item"] for item in shopping_list if not item["purchased"]]
remaining_cost = sum(item["price"] for item in shopping_list if not item["purchased"])
print(f"Still need to buy: {still_need}")
print(f"Remaining cost: {remaining_cost}")
```

```python
habits = [
    ("Drink water", True),
    ("Read 10 pages", False),
    ("Exercise", True),
    ("Sleep 8 hours", True),
    ("Meditate", False), 
]
def habit_report(habits_list):
    completed_count = sum(1 for habit, status in habits_list if status)
    total_count = len(habits_list)
    return {"completed": completed_count, "total": total_count}
for habit_name, completed in habits:
    status_str = "Done" if completed else "Not done"
    print(f"{habit_name}: {status_str}")
report = habit_report(habits)
print(report)
```
